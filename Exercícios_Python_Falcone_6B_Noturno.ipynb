Exercicio 3 - Calcular area de um circulo

---

*   Valor de PI: 3,14



```python
valorRaio = float(input("Digite o valor do raio: "))
pi = 3.14
area = pi * (valorRaio ** 2)
print("A area do circulo é: ", area)
```

Exercicio 8 - Verificar se a palavra é um palíndromo

---
*   Utilizar if e else para a conferencia do palíndromo
*   Usar o modelo [::-1] para inverter a palavra








```python
palavra = input("Digite uma palavra: ")
if palavra == palavra[::-1]:
  print("A palavra é um palíndromo")
else:
  print("A palavra não é um palíndromo")
```

Exercicio 9 - Realizar o fatorial de um numero fornecido

---
*   Aplicar fatorial

```python
numero = int(input("Digite um numero: "))
fatorial = 1
for i in range(1, numero + 1):
  fatorial *= i
print("O fatorial de", numero, "é", fatorial)
```

Exercicio 12 - Contar o numero de ocorrencias de cada caractere em uma string

---

*   Separar caracteres
*   Verificar a quantidade de vezes que cada caracter aparece

```python
texto = input("Digite um texto: ")
for caractere in set(texto):
  print(caractere, ":", texto.count(caractere))
```

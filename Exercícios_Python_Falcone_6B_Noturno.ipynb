Exercicio 2 - Somar 2 numeros

---

Exercicio 3 - Calcular area de um circulo

---

*   Valor de PI: 3,14



```python
valorRaio = float(input("Digite o valor do raio: "))
pi = 3.14
area = pi * (valorRaio ** 2)
print("A area do circulo é: ", area)
```

Exercicio 4 - Converter temperatura de C para F

---

*    Fórmula F = C * 9/5 + 32.



```python
celcius = float(input("Digite a temperatura em Celcius: "))
farenheit = celcius * 9/5 + 32
print("A temperatura em Farenheit é: ", farenheit)
```

Exercicio 5 - Realizar tabuada

---
*   Replicar o for i de java no python

```python
numero = int(input("Digite um numero: "))
for multiplicador in range(1, 11):
  print(numero, "x", multiplicador, "=", numero * multiplicador)
```

Exercicio 6 - Exibir numeros pares do 1 ao 20

---

*   Fazer um calculo com modulo usando if e else para exibir na tela


```python
for numero in range(1, 21):
  if numero % 2 == 0:
    print(numero)
```

Exercicio 7 - Verificar se o numero é par ou impar

---

*   Seguir o mesmo principio do exercicio 6 fazendo um if else para verificar se é par ou impar


```python
numero = int(input("Digite um numero: "))
if numero % 2 == 0:
  print("O numero é par")
else:
  print("O numero é impar")
```

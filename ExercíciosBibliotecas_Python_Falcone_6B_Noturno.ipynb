19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1,11, size=20)
array2 = np.random.randint(1,11, size=20)

produtoEscalar = np.dot(array1, array2)

normaArray1 = np.linalg.norm(array1)
normaArray2 = np.linalg.norm(array2)

angulo = np.arccos(produtoEscalar / (normaArray1 * normaArray2))

angulo = np.clip(angulo, -1.0, 1.0)

anguloRadianos = np.arccos(angulo)
anguloGraus = np.degrees(anguloRadianos)

print("Array 1: ", array1)
print("Array 2: ", array2)

print("Produto Escalar: ", produtoEscalar)

print("Angulo entre os arrays: ", anguloGraus)
```

Foi utilizado o método [np.random.randint] para gerar dois arrays de 20 números aleatórios entre 1 e 10.

Foi utilizado o método [np.dot] para calcular o produto escalar entre os arrays.

Foi utilizado o método [np.linalg.norm] para calcular a norma dos arrays.

Foi utilizado o método [np.arccos] para calcular o ângulo entre os arrays.

Foi utilizado o método [np.degrees] para converter o ângulo de radianos para graus.

2. Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

```python
import numpy as np
from scipy import integrate

def f(x, y):
    return x - y

xSpan = (0, 10)
y0 = [1.0]

xEval = np.linspace(0, 10, 11)

solucao = integrate.solve_ivp(f, xSpan, y0, t_eval=xEval)

for xVal, yVal in zip(solucao.t, solucao.y[0]):
    print("x =", xVal, " | y(x) =", yVal)
```

Foi utilizada a função [integrate.solve_ivp] para resolver numericamente a equação diferencial dy/dx = x - y.

Foi definida a condição inicial [y0 = 1.0] e o intervalo de 0 a 10.

Foi utilizado o método [np.linspace] para criar 11 pontos entre 0 e 10.

Foi utilizado o comando [zip] para percorrer os valores de x e y e exibir os resultados da solução.

4. Use a função de otimização de SciPy para maximizar f(x) =−xˆ2+ 4x+ 1.

```python
from scipy import optimize

def f(x):
  return -( -x[0]**2 + 4*x[0] + 1)

resultado = optimize.minimize(f, [0.0])
pontoMaximo = resultado.x[0]
valorMaximo = -resultado.fun

print("Ponto máximo (x): ", pontoMaximo)
print("Valor máximo: ", valorMaximo)
```

16. Resolva a equação diferencial de segunda ordem {d^2y}/{dx^2} + y = 0, com condições iniciais y(0) = 1 e y'(0) = 0

```python
import numpy as np
from scipy import integrate

def f(x, u):
  return [u[1], -u[0]]

xSpan = (0, 10)
condicoesIniciais = [1.0, 0.0]
xEval = np.linspace(0, 10, 11)

solucaoEDO = integrate.solve_ivp(f, xSpan, condicoesIniciais, t_eval=xEval)

for xVal, yVal in zip(solucaoEDO.t, solucaoEDO.y[0]):
  print("x =", xVal, " | y(x) =", yVal)
```

Foi utilizada a função [integrate.solve_ivp] para resolver numericamente a EDO `y'' = -y`.

Foram utilizadas as condições iniciais `[1.0, 0.0]` e o intervalo `[0, 10]`.

Foi utilizada [np.linspace] para gerar os pontos avaliados e [zip] para exibir os resultados.


17. Encontre a raiz cúbica de 27 utilizando SciPy

```python
from scipy import special
from scipy import optimize

def f(x):
  return x**3 - 27

raizEspecial = special.cbrt(27)
raizNumerica = optimize.fsolve(f, 1.0)[0]

print("Raiz (Funcao Especial): ", raizEspecial)
print("Raiz (Solucionador): ", raizNumerica)
```

Foi utilizada a função [special.cbrt] para calcular a raiz cúbica de 27.

Foi utilizada a função [optimize.fsolve] para encontrar numericamente a raiz da função `x³ - 27`.

Foram armazenados os resultados em [raizEspecial] e [raizNumerica].

19. Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos

```python
import numpy as np
from scipy import fft

t = np.linspace(0, 1, 256, endpoint=False)
sinal = np.sin(2 * np.pi * 5 * t) + np.cos(2 * np.pi * 10 * t)
transformadaSinal = fft.fft(sinal)

picosFrequencia = np.where(np.abs(transformadaSinal) > 10.0)[0]

print("Picos de frequência: ", picosFrequencia)
```

Foi utilizada a função [np.linspace] para criar os pontos do sinal no intervalo de 0 a 1.

Foi utilizada [fft.fft] para calcular a Transformada de Fourier do sinal.

Foi utilizada [np.where] para identificar os picos de frequência do sinal.

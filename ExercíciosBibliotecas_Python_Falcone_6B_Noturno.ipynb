9. Crie um array com 10 números aleatórios e normalize seus valores entre 0
e 1.


```python
import numpy as np

array = np.random.randint(1, 11, size=10)

print("Array original:")
print(array)

valorMin = np.min(array)
valorMax = np.max(array)

arrayNormalizado = (array - valorMin) / (valorMax - valorMin)

arrayNormalizado = np.round(arrayNormalizado, 2)

print("Array normalizado:")
print(arrayNormalizado)
```

Foi utilizado o metodo [np.random.randint] para gerar um array de numeros de 1 a 10 com o tamanho do array de 10 componentes

Foi utilizado os metodos [np.min] e [np.max] para acessar o menor e o maior numero do array

Foi utilizada a fórmula de normalização (x - mínimo) / (máximo - mínimo) para calcular o valor normalizado de cada número do array, transformando os valores para uma escala entre 0 e 1.

19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1,11, size=20)
array2 = np.random.randint(1,11, size=20)

produtoEscalar = np.dot(array1, array2)

normaArray1 = np.linalg.norm(array1)
normaArray2 = np.linalg.norm(array2)

angulo = np.arccos(produtoEscalar / (normaArray1 * normaArray2))

angulo = np.clip(angulo, -1.0, 1.0)

anguloRadianos = np.arccos(angulo)
anguloGraus = np.degrees(anguloRadianos)

print("Array 1: ", array1)
print("Array 2: ", array2)

print("Produto Escalar: ", produtoEscalar)

print("Angulo entre os arrays: ", anguloGraus)
```

Foi utilizado o método [np.random.randint] para gerar dois arrays de 20 números aleatórios entre 1 e 10.

Foi utilizado o método [np.dot] para calcular o produto escalar entre os arrays.

Foi utilizado o método [np.linalg.norm] para calcular a norma dos arrays.

Foi utilizado o método [np.arccos] para calcular o ângulo entre os arrays.

Foi utilizado o método [np.degrees] para converter o ângulo de radianos para graus.

1. Calcule a integral da função f(x) = x2 sin(x) no intervalo de 0 a π

```python
import numpy as np
from scipy import integrate

def f(x):
  return x**2 * np.sin(x)

resultado, erroEstimado = integrate.quad(f, 0, np.pi)

print("Resultado da integral: ", resultado)
print("Erro estimado: ", erroEstimado)
```

Foi utilizada a biblioteca [scipy.integrate] para realizar o cálculo da integral definida.

Foi utilizada a função [integrate.quad] para calcular a integral de x² · sen(x) no intervalo de 0 a π.

Foi utilizado o valor [erroEstimado] para apresentar uma estimativa do erro do cálculo.

2. Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

```python
import numpy as np
from scipy import integrate

def f(x, y):
    return x - y

xSpan = (0, 10)
y0 = [1.0]

xEval = np.linspace(0, 10, 11)

solucao = integrate.solve_ivp(f, xSpan, y0, t_eval=xEval)

for xVal, yVal in zip(solucao.t, solucao.y[0]):
    print("x =", xVal, " | y(x) =", yVal)
```

Foi utilizada a função [integrate.solve_ivp] para resolver numericamente a equação diferencial dy/dx = x - y.

Foi definida a condição inicial [y0 = 1.0] e o intervalo de 0 a 10.

Foi utilizado o método [np.linspace] para criar 11 pontos entre 0 e 10.

Foi utilizado o comando [zip] para percorrer os valores de x e y e exibir os resultados da solução.


3. Encontre as raízes da função f(x) = xˆ4 − 3xˆ3 + 2 utilizando SciPy.

```python
from scipy import optimize

def f(x):
  return x**4 - 3*x**3 + 2

raiz1 = optimize.fsolve(f, 1.0)[0]
raiz2 = optimize.fsolve(f, 3.0)[0]

print("Raiz 1: ", raiz1)
print("Raiz 2: ", raiz2)
```

Foi utilizada a função [optimize.fsolve] para encontrar numericamente as raízes da função x⁴ - 3x³ + 2.

Foi utilizado o valor [1.0] como aproximação inicial para encontrar a primeira raiz e [-1.0] para encontrar a segunda raiz.

Foi utilizado o comando [print] para exibir as raízes encontradas.

4. Use a função de otimização de SciPy para maximizar f(x) =−xˆ2+ 4x+ 1.

```python
from scipy import optimize

def f(x):
  return -( -x[0]**2 + 4*x[0] + 1)

resultado = optimize.minimize(f, [0.0])
pontoMaximo = resultado.x[0]
valorMaximo = -resultado.fun

print("Ponto máximo (x): ", pontoMaximo)
print("Valor máximo: ", valorMaximo)
```

Foi utilizada a função [optimize.minimize] para encontrar o ponto máximo da função [-x² + 4x + 1].

Foi utilizada a função negativa para transformar o problema de maximização em minimização.

Foi utilizado o comando [resultado.x] para obter o ponto máximo e [resultado.fun] para obter o valor máximo.

5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).

```python
import numpy as np
from scipy import fft

t = np.linspace(0, 1, 500, endpoint=False)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
transformadaSinal = fft.fft(sinal)

print("Primeiros 5 termos da transformada de Fourier: ", transformadaSinal[:5])
```

Foi utilizado o método [np.linspace] para criar 500 pontos no intervalo de 0 a 1.

Foi utilizada a função [np.sin] e [np.cos] para criar um sinal composto por duas ondas.

Foi utilizada a função [fft.fft] para calcular a transformada de Fourier do sinal.

Foi utilizado o comando [transformadaSinal[:5]] para exibir os 5 primeiros termos da transformada.

6. Encontre a matriz inversa de uma matriz 5x5 de números aleatórios.

```python
import numpy as np
from scipy import linalg

np.random.seed(42)
matrizAleatoria = np.random.rand(5,5)
matrizInversa = linalg.inv(matrizAleatoria)

print("Matriz original: ")
print(matrizAleatoria)

print("Matriz inversa: ")
print(matrizInversa)
```

7. Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0), (1, 2), (2, 3), (3, 5), (4, 4)

```python
import numpy as np
from scipy import interpolate

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 1, 2, 3, 4])

splineCubica = interpolate.CubicSpline(x, y)

valorInterpolado = splineCubica(2.5)

print("Valor interpolado em x = 2.5: ", valorInterpolado)
```

Foi utilizada a função [interpolate.CubicSpline] para criar uma interpolação cúbica a partir dos valores de x e y.

Foi utilizado o valor [2.5] para encontrar o valor interpolado correspondente.

Foi utilizado o comando [print] para exibir o resultado da interpolação.
